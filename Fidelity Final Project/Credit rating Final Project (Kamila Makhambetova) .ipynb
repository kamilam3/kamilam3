```python

import matplotlib.pyplot as plt
import seaborn as sns


def plot_company_rating(data, company, ycol):
    
    # Filter for 'Bunge Limited'
    data2 = data[data['Name'] == company]


    # Plot
    plt.figure(figsize=(10, 5))
    sns.scatterplot(data=data2, x='Date', y=ycol, hue='Rating', palette='viridis', s=100)
    
    # Annotate each point with the date
    for i in range(len(data2)):
        plt.annotate(data2['Date'].iloc[i].strftime('%m/%d/%Y'), 
                 (data2['Date'].iloc[i], data2[ycol].iloc[i]),
                 textcoords="offset points", xytext=(0,-10), ha='center', fontsize=9)
    
    if(data2['Rating_encoded'].min() != 0):
        plt.ylim(data2['Rating_encoded'].min() - 0.5, data2['Rating_encoded'].max() + 0.5)
    else:
        plt.ylim(data2['Rating_encoded'].min(), data2['Rating_encoded'].max() + 0.5)
    plt.xlabel("Date")
    plt.ylabel(ycol)
    plt.title(ycol+ " vs Time for " + company + " over Aug 2005 - Dec 2016")
    plt.xticks(rotation=45)
    plt.legend(title="Rating")
    plt.show()

    
```

```python

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt




def freq_graph(data, col, rotation):
    
    if data[col].dtype != 'object':  
        data[col] = data[col].astype(str)
    
    category_counts = data[col].value_counts()
    
    sns.barplot(x=category_counts.index, y=category_counts.values, palette="pastel")
    
    # Add labels and title
    plt.title('Frequency of ' + col, fontsize=14)
    plt.xlabel(col, fontsize=12)
    plt.ylabel('Frequency', fontsize=12)
    
    if(rotation==1):
        
        # Rotate the x-axis labels to avoid overlap
        plt.xticks(rotation=45, ha='right')

    # Annotate the bars with the frequency count
    for i, count in enumerate(category_counts.values):
        plt.text(i, count + 0.5, str(count), ha='center', fontsize=10)
        
    # Show the plot
    plt.tight_layout()
    plt.show()
    
    
```

```python
def count_outliers(data, col):

    Q1 = data[col].quantile(0.25)
    Q3 = data[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = data[(data[col] < lower_bound) | (data[col] > upper_bound)]
    
    num_outliers = len(outliers)
    percentage_outliers = (num_outliers / len(data)) * 100
    
    print(f"Number of outliers in {col} : {num_outliers} ({percentage_outliers:.2f}%)")
    #print(f"Percentage of outliers in {col} : {percentage_outliers:.2f}%")
```

```python

def count_outliers2(data, cols):
    outlier_summary = []

    for col in cols:  
        Q1 = data[col].quantile(0.25)
        Q3 = data[col].quantile(0.75)
        IQR = Q3 - Q1
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR
        
        outliers = data[(data[col] < lower_bound) | (data[col] > upper_bound)]
        
        num_outliers = len(outliers)
        percentage_outliers = (num_outliers / len(data)) * 100
        
        outlier_summary.append({'Column': col, 'Num Outliers': num_outliers, 'Percentage Outliers': round(percentage_outliers, 2)})

    return pd.DataFrame(outlier_summary)
```

```python
def count_row_outliers(data):

    outlier_counts = pd.Series(0, index=data.index)
    
    for col in data.columns:
        Q1 = data[col].quantile(0.25)
        Q3 = data[col].quantile(0.75)
        IQR = Q3 - Q1
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR
        
        outlier_mask = (data[col] < lower_bound) | (data[col] > upper_bound)
        outlier_counts += outlier_mask.astype(int)
    
    return outlier_counts
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

def draw_boxplots_without_outliers(data, predictors, class_col):

    
    # Loop through each predictor column
    for predictor in predictors:
        # Calculate Q1, Q3, and IQR for each predictor
        Q1 = data[predictor].quantile(0.25)
        Q3 = data[predictor].quantile(0.75)
        IQR = Q3 - Q1
        
        # Define the outlier bounds
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR
        
        # Filter out the outliers
        filtered_data = data[(data[predictor] >= lower_bound) & (data[predictor] <= upper_bound)]
        
        # Create the boxplot for the predictor column with the category column
        plt.figure(figsize=(8, 6))
        sns.boxplot(x=class_col, y=predictor, data=filtered_data, palette='Set2')
        
        # Set labels and title
        plt.xlabel(class_col)
        plt.ylabel(predictor)
        plt.title(f"Boxplot of {predictor} by {class_col} (Outliers Excluded)")
        
        # Show the plot
        plt.show()
```

# Substitute outliers with lower/ upper bound

```python
def cap_outliers(data_inp, cols):
    
    data = data_inp.copy()
    for col in cols:  # Process only numeric columns
        Q1 = data[col].quantile(0.25)
        Q3 = data[col].quantile(0.75)
        IQR = Q3 - Q1
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        data[col] = data[col].clip(lower=lower_bound, upper=upper_bound)  # Capping outliers

    return data

# Example usage:
# df = cap_outliers(df)
```

```python
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn import preprocessing, metrics
from sklearn.model_selection import train_test_split
from sklearn.ensemble import GradientBoostingClassifier, RandomForestClassifier
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis, QuadraticDiscriminantAnalysis
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression
import xgboost as xgb
from sklearn.metrics import f1_score
from sklearn.utils.class_weight import compute_sample_weight
from sklearn.metrics import make_scorer, accuracy_score, classification_report
from sklearn.preprocessing import LabelEncoder
from sklearn import metrics, preprocessing
import numpy as np



def class_models(data_inp, X_cols, y_col, test_size, num_class):
    
    data = data_inp.copy()
    
    le = preprocessing.LabelEncoder()
    le.fit(data.Sector)
    data.Sector = le.transform(data.Sector)  # Encode sector
    
    # Split dataset into training and testing while maintaining category proportions
    X = data[X_cols]  # Features
    y = data[y_col]  # Target
    
    X_train, X_test, y_train, y_test = train_test_split(X, y, 
                    test_size=test_size, stratify=y, random_state=123)
    
    # Compute class distribution for LDA
    classes, counts = np.unique(y_train, return_counts=True)
    class_priors = counts / np.sum(counts)  # Compute priors based on training data
    
    # Initialize models
    models = {
        "Gradient Boosting": GradientBoostingClassifier(random_state=123),
        "Random Forest": RandomForestClassifier(n_estimators=200, class_weight='balanced', random_state=123),
        "Linear Discriminant Analysis": LinearDiscriminantAnalysis(priors=class_priors),
        "Logistic Regression": LogisticRegression(class_weight='balanced', random_state=123,
                                                  multi_class='multinomial', solver='newton-cg')
    }
    
    accuracies = {}
    sample_weights = compute_sample_weight(class_weight="balanced", y=y_train)

    # Train and evaluate each model
    for name, model in models.items():
        
        if(name == "Gradient Boosting"):
            model.fit(X_train, y_train, sample_weight=sample_weights)
        
        else:
            model.fit(X_train, y_train)

        y_pred = model.predict(X_test)
        accuracy = metrics.accuracy_score(y_test, y_pred)
        accuracies[name] = accuracy
        print(f"{name} Accuracy: {accuracy:.4f}")

    # Train KNN models with k from 1 to 10 and find the best k
    best_knn_k = 1
    best_knn_accuracy = 0

    for k in range(1, 11):
        knn_model = KNeighborsClassifier(n_neighbors=k, weights='distance')
        knn_model.fit(X_train, y_train)
        y_pred_knn = knn_model.predict(X_test)
        accuracy_knn = metrics.accuracy_score(y_test, y_pred_knn)
        
        if accuracy_knn > best_knn_accuracy:
            best_knn_accuracy = accuracy_knn
            best_knn_k = k
    
    # Store the best KNN accuracy
    accuracies[f"KNN (k={best_knn_k})"] = best_knn_accuracy
    print(f"Best KNN Accuracy ({best_knn_k} neighbors): {best_knn_accuracy:.4f}")

    # Convert to DataFrame and sort by accuracy
    df_accuracy = pd.DataFrame(list(accuracies.items()), columns=['Model', 'Accuracy'])
    df_accuracy = df_accuracy.sort_values(by='Accuracy', ascending=False)

    # Plot the accuracies
    plt.figure(figsize=(12, 6))
    ax = sns.barplot(x='Accuracy', y='Model', data=df_accuracy, palette='viridis')
    
    # Add percentage labels on bars
    for index, value in enumerate(df_accuracy["Accuracy"]):
        ax.text(value + 0.01, index, f"{value:.2%}", va='center', fontsize=12, fontweight='bold', color='black')

    plt.xlabel("Accuracy")
    plt.ylabel("Model")
    plt.title("Comparison of Model Accuracies for " + y_col)
    plt.xlim(0, 1)  # Accuracy ranges from 0 to 1
    plt.grid(axis='x', linestyle='--', alpha=0.7)
    plt.show()

    return df_accuracy  





def class_models_accur_confus_matrix(data_inp, X_cols, y_col, test_size, num_class):
    data = data_inp.copy()
    
    le = preprocessing.LabelEncoder()
    le.fit(data.Sector)
    data.Sector = le.transform(data.Sector)  # Encode sector
    
    # Split dataset into training and testing while maintaining category proportions
    X = data[X_cols]  # Features
    y = data[y_col]  # Target
    
    X_train, X_test, y_train, y_test = train_test_split(X, y, 
                    test_size=test_size, stratify=y, random_state=123)
    
    # Compute class distribution for LDA
    classes, counts = np.unique(y_train, return_counts=True)
    class_priors = counts / np.sum(counts)  # Compute priors based on training data
    
    # Initialize models
    models = {
        "Gradient Boosting": GradientBoostingClassifier(random_state=123),
        "Random Forest": RandomForestClassifier(n_estimators=200, class_weight='balanced', random_state=123),
        "Linear Discriminant Analysis": LinearDiscriminantAnalysis(priors=class_priors),
        "Logistic Regression": LogisticRegression(class_weight='balanced', random_state=123,
                                                  multi_class='multinomial', solver='newton-cg')
    }
    
    accuracies = {}
    sample_weights = compute_sample_weight(class_weight="balanced", y=y_train)

    # Train and evaluate each model
    for name, model in models.items():
        
        if name == "Gradient Boosting":
            model.fit(X_train, y_train, sample_weight=sample_weights)
        else:
            model.fit(X_train, y_train)

        y_pred = model.predict(X_test)
        accuracy = metrics.accuracy_score(y_test, y_pred)
        accuracies[name] = accuracy
        
        print(f"{name} Accuracy: {accuracy:.4f}")
        print(f"{name} Classification Report:\n{metrics.classification_report(y_test, y_pred)}")
        print(f"{name} Confusion Matrix:\n{metrics.confusion_matrix(y_test, y_pred)}\n")
    
    # Train KNN models with k from 1 to 10 and find the best k
    best_knn_k = 1
    best_knn_accuracy = 0

    for k in range(1, 11):
        knn_model = KNeighborsClassifier(n_neighbors=k, weights='distance')
        knn_model.fit(X_train, y_train)
        y_pred_knn = knn_model.predict(X_test)
        accuracy_knn = metrics.accuracy_score(y_test, y_pred_knn)
        
        if accuracy_knn > best_knn_accuracy:
            best_knn_accuracy = accuracy_knn
            best_knn_k = k
    
    # Store the best KNN accuracy
    accuracies[f"KNN (k={best_knn_k})"] = best_knn_accuracy
    print(f"Best KNN Accuracy ({best_knn_k} neighbors): {best_knn_accuracy:.4f}")
    
    # Convert to DataFrame and sort by accuracy
    df_accuracy = pd.DataFrame(list(accuracies.items()), columns=['Model', 'Accuracy'])
    df_accuracy = df_accuracy.sort_values(by='Accuracy', ascending=False)
    
    # Plot the accuracies
    plt.figure(figsize=(12, 6))
    ax = sns.barplot(x='Accuracy', y='Model', data=df_accuracy, palette='viridis')
    
    # Add percentage labels on bars
    for index, value in enumerate(df_accuracy["Accuracy"]):
        ax.text(value + 0.01, index, f"{value:.2%}", va='center', fontsize=12, fontweight='bold', color='black')

    plt.xlabel("Accuracy")
    plt.ylabel("Model")
    plt.title("Comparison of Model Accuracies for " + y_col)
    plt.xlim(0, 1)  # Accuracy ranges from 0 to 1
    plt.grid(axis='x', linestyle='--', alpha=0.7)
    plt.show()
    
    return df_accuracy



def class_models_f1(data_inp, X_cols, y_col, test_size, num_class):
    
    data = data_inp.copy()
    
    le = preprocessing.LabelEncoder()
    le.fit(data.Sector)
    data.Sector = le.transform(data.Sector)  # Encode sector

    
    # Split dataset into training and testing while maintaining category proportions
    X = data[X_cols]  # Features
    y = data[y_col]  # Target
    
    X_train, X_test, y_train, y_test = train_test_split(X, y, 
                    test_size=test_size, stratify=y, random_state=123)
    
    # Compute class distribution for LDA
    classes, counts = np.unique(y_train, return_counts=True)
    class_priors = counts / np.sum(counts)  # Compute priors based on training data
    
    # Initialize models
    models = {

        "Gradient Boosting": GradientBoostingClassifier(random_state=123),
        "Random Forest": RandomForestClassifier(n_estimators=200, class_weight='balanced', random_state=123),
        "Linear Discriminant Analysis": LinearDiscriminantAnalysis(priors=class_priors),
        "Logistic Regression": LogisticRegression(class_weight='balanced', random_state=123,
                                                  multi_class='multinomial', solver='newton-cg')
    }
    
    f1_scores = {}
    sample_weights = compute_sample_weight(class_weight="balanced", y=y_train)
    

    # Train and evaluate each model
    for name, model in models.items():
        
        if(name == "Gradient Boosting"):
            model.fit(X_train, y_train, sample_weight=sample_weights)
        
        else:
            model.fit(X_train, y_train)
        
        y_pred = model.predict(X_test)
        f1 = f1_score(y_test, y_pred, average='weighted')
        f1_scores[name] = f1
        
        print(f"{name} F1 Score: {f1:.4f}")
        
        

    # Train KNN models with k from 1 to 10 and find the best k
    best_knn_k = 1
    best_knn_f1 = 0

    for k in range(1, 11):
        knn_model = KNeighborsClassifier(n_neighbors=k, weights='distance')
        knn_model.fit(X_train, y_train)
        y_pred_knn = knn_model.predict(X_test)
        f1_knn = f1_score(y_test, y_pred_knn, average='weighted')
        
        if f1_knn > best_knn_f1:
            best_knn_f1 = f1_knn
            best_knn_k = k
    
    # Store the best KNN F1 score
    f1_scores[f"KNN (k={best_knn_k})"] = best_knn_f1
    print(f"Best KNN F1 Score ({best_knn_k} neighbors): {best_knn_f1:.4f}")

    # Convert to DataFrame and sort by F1 score
    df_f1 = pd.DataFrame(list(f1_scores.items()), columns=['Model', 'F1 Score'])
    df_f1 = df_f1.sort_values(by='F1 Score', ascending=False)

    # Plot the F1 scores
    plt.figure(figsize=(12, 6))
    ax = sns.barplot(x='F1 Score', y='Model', data=df_f1, palette='viridis')
    
    # Add percentage labels on bars
    for index, value in enumerate(df_f1["F1 Score"]):
        ax.text(value + 0.01, index, f"{value:.2f}", va='center', fontsize=12, fontweight='bold', color='black')

    plt.xlabel("F1 Score")
    plt.ylabel("Model")
    plt.title("Comparison of Model F1 Scores for " + y_col)
    plt.xlim(0, 1)  # F1 score ranges from 0 to 1
    plt.grid(axis='x', linestyle='--', alpha=0.7)
    plt.show()

    return df_f1


def class_models_macro_avg(data_inp, X_cols, y_col, test_size, num_class):
    
    data = data_inp.copy()
    
    # Encode categorical variables
    le = preprocessing.LabelEncoder()
    le.fit(data.Sector)
    data.Sector = le.transform(data.Sector)  # Encode sector

    
    # Split dataset into training and testing while maintaining category proportions
    X = data[X_cols]  # Features
    y = data[y_col]  # Target
    
    X_train, X_test, y_train, y_test = train_test_split(X, y, 
                    test_size=test_size, stratify=y, random_state=123)
    
    # Compute class distribution for LDA
    classes, counts = np.unique(y_train, return_counts=True)
    class_priors = counts / np.sum(counts)  # Compute priors based on training data

    
    # Initialize models
    models = {
        "Gradient Boosting": GradientBoostingClassifier(random_state=123),
        "Random Forest": RandomForestClassifier(n_estimators=200, class_weight='balanced', random_state=123),
        "Linear Discriminant Analysis": LinearDiscriminantAnalysis(priors=class_priors),
        "Logistic Regression": LogisticRegression(class_weight='balanced', random_state=123,
                                                  multi_class='multinomial', solver='newton-cg')
    }
    
    accuracies = {}
    
    sample_weights = compute_sample_weight(class_weight="balanced", y=y_train)
    
    

    # Train and evaluate each model
    for name, model in models.items():
        
        if(name == "Gradient Boosting"):
            model.fit(X_train, y_train, sample_weight=sample_weights)
        
        else:
            model.fit(X_train, y_train)
            
        y_pred = model.predict(X_test)
        
        
        # Compute macro-average accuracy
        macro_avg_accuracy = metrics.balanced_accuracy_score(y_test, y_pred)
        accuracies[name] = macro_avg_accuracy
        print(f"{name} Macro-Average Accuracy: {macro_avg_accuracy:.4f}")

    # Train KNN models with k from 1 to 10 and find the best k
    best_knn_k = 1
    best_knn_accuracy = 0

    for k in range(1, 11):
        knn_model = KNeighborsClassifier(n_neighbors=k, weights='distance')
        knn_model.fit(X_train, y_train)
        y_pred_knn = knn_model.predict(X_test)
        
        # Compute macro-average accuracy for KNN
        accuracy_knn = metrics.balanced_accuracy_score(y_test, y_pred_knn)
        
        if accuracy_knn > best_knn_accuracy:
            best_knn_accuracy = accuracy_knn
            best_knn_k = k
    
    # Store the best KNN accuracy
    accuracies[f"KNN (k={best_knn_k})"] = best_knn_accuracy
    print(f"Best KNN Macro-Average Accuracy ({best_knn_k} neighbors): {best_knn_accuracy:.4f}")

    # Convert to DataFrame and sort by accuracy
    df_accuracy = pd.DataFrame(list(accuracies.items()), columns=['Model', 'Macro Avg Accuracy'])
    df_accuracy = df_accuracy.sort_values(by='Macro Avg Accuracy', ascending=False)

    # Plot the accuracies
    plt.figure(figsize=(12, 6))
    ax = sns.barplot(x='Macro Avg Accuracy', y='Model', data=df_accuracy, palette='viridis')
    
    # Add percentage labels on bars
    for index, value in enumerate(df_accuracy["Macro Avg Accuracy"]):
        ax.text(value + 0.01, index, f"{value:.2%}", va='center', fontsize=12, fontweight='bold', color='black')

    plt.xlabel("Macro-Average Accuracy")
    plt.ylabel("Model")
    plt.title("Comparison of Model Macro Avg Accuracies for " + y_col)
    plt.xlim(0, 1)  # Accuracy ranges from 0 to 1
    plt.grid(axis='x', linestyle='--', alpha=0.7)
    plt.show()

    return df_accuracy  



def class_models_macro_f1(data_inp, X_cols, y_col, test_size, num_class):
    
    data = data_inp.copy()
    
    le = preprocessing.LabelEncoder()
    le.fit(data.Sector)
    data.Sector = le.transform(data.Sector)  # Encode sector

    
    # Split dataset into training and testing while maintaining category proportions
    X = data[X_cols]  # Features
    y = data[y_col]  # Target
    
    X_train, X_test, y_train, y_test = train_test_split(X, y, 
                    test_size=test_size, stratify=y, random_state=123)
    
    # Compute class distribution for LDA
    classes, counts = np.unique(y_train, return_counts=True)
    class_priors = counts / np.sum(counts)  # Compute priors based on training data
    
    # Initialize models
    models = {
        "Gradient Boosting": GradientBoostingClassifier(random_state=123),
        "Random Forest": RandomForestClassifier(n_estimators=200, class_weight='balanced', random_state=123),
        "Linear Discriminant Analysis": LinearDiscriminantAnalysis(priors=class_priors),
        "Logistic Regression": LogisticRegression(class_weight='balanced', random_state=123, multi_class='multinomial', solver='newton-cg')
    }
    
   
    f1_scores = {}
    sample_weights = compute_sample_weight(class_weight="balanced", y=y_train)

    # Train and evaluate each model using MACRO F1
    for name, model in models.items():
        
        if(name == "Gradient Boosting"):
            model.fit(X_train, y_train, sample_weight=sample_weights)
        
        else:
            model.fit(X_train, y_train)
            

        y_pred = model.predict(X_test)
        
        #Compute macro avg F1
        f1 = f1_score(y_test, y_pred, average='macro')  
        f1_scores[name] = f1
        print(f"{name} Macro F1 Score: {f1:.4f}")

    # Train KNN models with k from 1 to 10 and find the best k
    best_knn_k = 1
    best_knn_f1 = 0

    for k in range(1, 11):
        knn_model = KNeighborsClassifier(n_neighbors=k, weights='distance')
        knn_model.fit(X_train, y_train)
        y_pred_knn = knn_model.predict(X_test)
        
        f1_knn = f1_score(y_test, y_pred_knn, average='macro')  
        
        if f1_knn > best_knn_f1:
            best_knn_f1 = f1_knn
            best_knn_k = k
    
    # Store the best KNN Macro F1 score
    f1_scores[f"KNN (k={best_knn_k})"] = best_knn_f1
    print(f"Best KNN Macro F1 Score ({best_knn_k} neighbors): {best_knn_f1:.4f}")

    # Convert to DataFrame and sort by F1 score
    df_f1 = pd.DataFrame(list(f1_scores.items()), columns=['Model', 'F1 Score'])
    df_f1 = df_f1.sort_values(by='F1 Score', ascending=False)

    # Plot the F1 scores
    plt.figure(figsize=(12, 6))
    ax = sns.barplot(x='F1 Score', y='Model', data=df_f1, palette='viridis')
    
    # Add percentage labels on bars
    for index, value in enumerate(df_f1["F1 Score"]):
        ax.text(value + 0.01, index, f"{value:.2f}", va='center', fontsize=12, fontweight='bold', color='black')

    plt.xlabel("F1 Score")
    plt.ylabel("Model")
    plt.title("Comparison of Model Macro F1 Scores for " + y_col)
    plt.xlim(0, 1)  # F1 score ranges from 0 to 1
    plt.grid(axis='x', linestyle='--', alpha=0.7)
    plt.show()

    return df_f1
```

```python

# def compare_bar_charts(df1, df2, title,  y_col,  label1, label2, xlim,  color1='blue', color2='red'):
#     """
#     Creates a horizontal bar chart comparing values from two datasets.
    
#     Parameters:
#     df1, df2: DataFrames with the same index (categories) and one column with numerical values.
#     title: Title of the chart (default: 'Comparison of Scores')
#     color1, color2: Colors for the bars (default: blue and red)
#     """
#     categories = df1['Model']
#     values1 = df1.iloc[:, 1].values  # Extract first column as values
#     values2 = df2.iloc[:, 1].values  # Extract first column as values
    
#     y_pos = np.arange(len(categories))
    
#     plt.figure(figsize=(8, 6))
    
#     # Plot bars for both datasets
#     plt.barh(y_pos + 0.2, values2, height=0.4, color=color2, alpha=0.7, label=label2)
#     plt.barh(y_pos - 0.2, values1, height=0.4, color=color1, alpha=0.7, label=label1)

    
#     # Add labels
#     plt.yticks(y_pos, categories)
#     plt.xlabel('Score')
#     plt.title(title +' for '+y_col)
    
#     # Add value labels to bars
#     for i in range(len(categories)):
#         plt.text(values1[i] + 0.01, y_pos[i] - 0.2, f"{values1[i]:.2f}", va='center', fontsize=10, fontweight='bold', color='black')
#         plt.text(values2[i] + 0.01, y_pos[i] + 0.2, f"{values2[i]:.2f}", va='center', fontsize=10, fontweight='bold', color='black')
    
#     plt.xlim(0, xlim)
#     plt.legend()
#     plt.grid(axis='x', linestyle='--', alpha=0.7)
#     plt.show()

# Example usage:
# compare_bar_charts(df1, df2)



import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

def compare_bar_charts(df1, df2, title, y_col, label1, label2, xlim, color1='blue', color2='red'):
    """
    Creates a horizontal bar chart comparing values from two datasets, aligning them by the first column.

    Parameters:
    df1, df2: DataFrames with a common first column (e.g., 'Model'), which will be used for merging.
    title: Title of the chart.
    y_col: Column name being compared.
    label1, label2: Labels for the datasets.
    xlim: Maximum value for the x-axis.
    color1, color2: Colors for the bars (default: blue and red).
    """
    
    # Clean 'Model' column to remove any text after 'KNN'
    df1_cleaned = df1.copy()
    df2_cleaned = df2.copy()

    df1_cleaned['Model'] = df1_cleaned['Model'].replace(r'KNN.*', 'KNN', regex=True)
    df2_cleaned['Model'] = df2_cleaned['Model'].replace(r'KNN.*', 'KNN', regex=True)

    # Merge datasets on the cleaned 'Model' column
    merged_df = pd.merge(df1_cleaned, df2_cleaned, on=df1.columns[0], suffixes=('_1', '_2'))


    categories = merged_df.iloc[:, 0]  # First column (e.g., 'Model')
    values1 = merged_df.iloc[:, 1].values  # Second column (values from df1)
    values2 = merged_df.iloc[:, 2].values  # Third column (values from df2)

    y_pos = np.arange(len(categories))

    plt.figure(figsize=(8, 6))

    # Plot bars for both datasets
    plt.barh(y_pos + 0.2, values2, height=0.4, color=color2, alpha=0.7, label=label2)
    plt.barh(y_pos - 0.2, values1, height=0.4, color=color1, alpha=0.7, label=label1)

    # Add labels
    plt.yticks(y_pos, categories)
    plt.xlabel('Macro Avg F1 Score')
    plt.title(f"{title} for {y_col}")

    # Add value labels to bars
    for i in range(len(categories)):
        plt.text(values1[i] + 0.01, y_pos[i] - 0.2, f"{values1[i]:.2f}", va='center', fontsize=10, fontweight='bold', color='black')
        plt.text(values2[i] + 0.01, y_pos[i] + 0.2, f"{values2[i]:.2f}", va='center', fontsize=10, fontweight='bold', color='black')

    plt.xlim(0, xlim)
    plt.legend()
    plt.grid(axis='x', linestyle='--', alpha=0.7)
    plt.show()

```

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from statsmodels.stats.outliers_influence import variance_inflation_factor

def cor(data, cols):
    correlation_matrix = data[cols].corr()
    print(correlation_matrix)
    
    plt.figure(figsize=(14, 10))
    sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f')
    #plt.savefig("correlation_heatmap.png", dpi=300, bbox_inches='tight')  # Save as PNG
    plt.show()
    
    
    
def vif(data, cols, save_path=None,font = 10):
                
    X = data[cols]
    vif_data = pd.DataFrame()
    vif_data["feature"] = X.columns
    vif_data["VIF"] = [round(variance_inflation_factor(X.values, i), 2) for i in range(len(X.columns))]
    vif_data = vif_data.sort_values(by='VIF', ascending=True).reset_index(drop=True)
    
    if save_path:

        fig, ax = plt.subplots(figsize=(6, len(cols) * 0.4))
        ax.axis('tight')
        ax.axis('off')
        table = ax.table(cellText=vif_data.values, colLabels=vif_data.columns, cellLoc='center', loc='center')
        table.auto_set_font_size(False)
        table.set_fontsize(font)
        plt.savefig(save_path, bbox_inches='tight', dpi=300)  # Save as PNG
    return vif_data

```

```python

from mlxtend.feature_selection import SequentialFeatureSelector
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.metrics import classification_report, make_scorer
from sklearn.utils.class_weight import compute_sample_weight
from sklearn.model_selection import train_test_split
from sklearn import preprocessing
import numpy as np

def backward_elimin_gb(data_inp, X_cols, y_col, test_size, num_class):
    
    data = data_inp.copy()

    le = preprocessing.LabelEncoder()
    le.fit(data.Sector)
    data.Sector = le.transform(data.Sector)  # Encode sector

    # Split dataset into training and testing while maintaining category proportions
    X = data[X_cols]  # Features
    y = data[y_col]  # Target
    
    X_train, X_test, y_train, y_test = train_test_split(X, y, 
                    test_size=test_size, stratify=y, random_state=123)
    
    # Custom function to calculate macro-averaged F1-score
    def macro_avg_f1(y_true, y_pred):
        report = classification_report(y_true, y_pred, output_dict=True)
        return report['macro avg']['f1-score']

    # Define custom scoring metric
    macro_scorer = make_scorer(macro_avg_f1)

    # Define a wrapper model that applies sample weights correctly
    class WeightedGBClassifier(GradientBoostingClassifier):
        def fit(self, X, y, **kwargs):
            sample_weights = compute_sample_weight(class_weight="balanced", y=y)  # Compute per fold
            return super().fit(X, y, sample_weight=sample_weights, **kwargs)

    # Initialize Gradient Boosting Model with sample weights
    gb_model = WeightedGBClassifier(random_state=123)

    # Backward Feature Selection using Gradient Boosting with sample weights
    backward_feature_selector = SequentialFeatureSelector(
        estimator=gb_model,  # Use wrapped model
        k_features=(1, X_train.shape[1]),  # Select between 1 and all features
        forward=False,  # Backward elimination
        floating=False,  # Standard backward elimination
        verbose=2,
        scoring=macro_scorer,  # Optimize for macro-averaged F1-score
        cv=5  # Cross-validation for stability
    )

    # Fit feature selector
    backward_feature_selector.fit(X_train, y_train)

    # Get selected feature names
    selected_features = list(backward_feature_selector.k_feature_names_)
    print("Optimal feature set:", selected_features)

    return selected_features




def forward_select_gb(data_inp, X_cols, y_col, test_size, num_class):
    
    data = data_inp.copy()

    le = preprocessing.LabelEncoder()
    le.fit(data.Sector)
    data.Sector = le.transform(data.Sector)  # Encode sector

    # Split dataset into training and testing while maintaining category proportions
    X = data[X_cols]  # Features
    y = data[y_col]  # Target
    
    X_train, X_test, y_train, y_test = train_test_split(X, y, 
                    test_size=test_size, stratify=y, random_state=123)
    
    # Custom function to calculate macro-averaged F1-score
    def macro_avg_f1(y_true, y_pred):
        report = classification_report(y_true, y_pred, output_dict=True)
        return report['macro avg']['f1-score']

    # Define custom scoring metric
    macro_scorer = make_scorer(macro_avg_f1)

    # Define a wrapper model that applies sample weights correctly
    class WeightedGBClassifier(GradientBoostingClassifier):
        def fit(self, X, y, **kwargs):
            sample_weights = compute_sample_weight(class_weight="balanced", y=y)  # Compute per fold
            return super().fit(X, y, sample_weight=sample_weights, **kwargs)

    # Initialize Gradient Boosting Model with sample weights
    gb_model = WeightedGBClassifier(random_state=123)

    # Forward Feature Selection using Gradient Boosting with sample weights
    forward_feature_selector = SequentialFeatureSelector(
        estimator=gb_model,  # Use wrapped model
        k_features=(1, X_train.shape[1]),  # Select between 1 and all features
        forward=True,  # Forward selection
        floating=False,  # Standard forward selection
        verbose=2,
        scoring=macro_scorer,  # Optimize for macro-averaged F1-score
        cv=5  # Cross-validation for stability
    )

    # Fit feature selector
    forward_feature_selector.fit(X_train, y_train)

    # Get selected feature names
    selected_features = list(forward_feature_selector.k_feature_names_)
    print("Optimal feature set:", selected_features)

    return selected_features



```

## Parameter tunning ML models


```python
from sklearn.model_selection import GridSearchCV
from sklearn.preprocessing import LabelEncoder
from sklearn.ensemble import GradientBoostingClassifier, RandomForestClassifier
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import f1_score, make_scorer
from sklearn.utils.class_weight import compute_sample_weight
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split

def class_models_macro_f1_param_tunning(data_inp, X_cols, y_col, test_size):
    
    data = data_inp.copy()
    
    # Encode sector
    le = preprocessing.LabelEncoder()
    le.fit(data.Sector)
    data.Sector = le.transform(data.Sector)  
    
    # Split dataset into training and testing while maintaining category proportions
    X = data[X_cols]  # Features
    y = data[y_col]  # Target
    
    X_train, X_test, y_train, y_test = train_test_split(X, y, 
                    test_size=test_size, stratify=y, random_state=123)
    
    #print('y_test classes: ', np.unique(y_test))
    #print('y_train classes: ', np.unique(y_train))
    
    # Compute sample weights
    sample_weights = compute_sample_weight(class_weight="balanced", y=y_train)
    
    # Define models and parameter grids
    param_grids = {
        "Gradient Boosting": {
            "model": GradientBoostingClassifier(random_state=123),
            "params": {"n_estimators": [100, 200,300, 400, 500],
                       "learning_rate": [0.01, 0.05, 0.1, 0.2, 0.3], "max_depth": [3, 4, 5, 6, 7, 8, 9]}
        },
        "Random Forest": {
            "model": RandomForestClassifier(class_weight='balanced', random_state=123),
            "params": {"n_estimators": [100, 200, 300], "max_depth": [None, 10, 20], "min_samples_split": [2,5,10]}
        },
        "Linear Discriminant Analysis": {
            "model": LinearDiscriminantAnalysis(),
            "params": {"solver": ["svd", "lsqr"]}
        },
        "Logistic Regression": {
            "model": LogisticRegression(class_weight='balanced', random_state=123, multi_class='multinomial', solver='newton-cg'),
            "params": {"C": [0.1, 1, 10]}
        },
        "KNN": {
            "model": KNeighborsClassifier(weights='distance'),
            "params": {"n_neighbors": list(range(1, 20))}
        }
    }
    
    f1_scorer = make_scorer(f1_score, average='macro', pos_label=None)
    best_params_list = []
    
    # Perform GridSearchCV for each model
    for name, config in param_grids.items():
        grid_search = GridSearchCV(config["model"], config["params"], scoring=f1_scorer, cv=5, n_jobs=-1, verbose=1)
        
        if name == "Gradient Boosting":
            #print("Unique classes in y_train:", np.unique(y_train))
            #print("Unique classes in y_test:", np.unique(y_test))
            grid_search.fit(X_train, y_train, sample_weight=sample_weights)
        else:
            #print("Unique classes in y_train:", np.unique(y_train))
            #print("Unique classes in y_test:", np.unique(y_test))
            grid_search.fit(X_train, y_train)
        
        best_params_list.append({"Model": name, "Best Parameters": grid_search.best_params_, "Best F1 Score": grid_search.best_score_})
    
    df_best_params = pd.DataFrame(best_params_list)
    df_best_params = df_best_params.sort_values(by='Best F1 Score', ascending=False)
    
    # Plot the F1 scores
    plt.figure(figsize=(12, 6))
    ax = sns.barplot(x='Best F1 Score', y='Model', data=df_best_params, palette='viridis')
    
    for index, value in enumerate(df_best_params["Best F1 Score"]):
        ax.text(value + 0.01, index, f"{value:.2f}", va='center', fontsize=12, fontweight='bold', color='black')
    
    plt.xlabel("F1 Score")
    plt.ylabel("Model")
    plt.title("Comparison of Best Model Macro F1 Scores for " + y_col)
    plt.xlim(0, 1)
    plt.grid(axis='x', linestyle='--', alpha=0.7)
    plt.show()
    
    return df_best_params






def class_models_macro_f1_params(data_inp, X_cols, y_col, test_size, model_params):
    data = data_inp.copy()
    
    # Encode sector
    le = preprocessing.LabelEncoder()
    le.fit(data.Sector)
    data.Sector = le.transform(data.Sector)  
    
    # Split dataset into training and testing while maintaining category proportions
    X = data[X_cols]  # Features
    y = data[y_col]  # Target
    
    X_train, X_test, y_train, y_test = train_test_split(X, y, 
                    test_size=test_size, stratify=y, random_state=123)
    
    # Compute sample weights
    sample_weights = compute_sample_weight(class_weight="balanced", y=y_train)
    
    f1_scores = {}
    
    # Train and evaluate each model using provided parameters
    for name, params in model_params.items():
        model = params["model"](**params["params"])
        
        if name == "Gradient Boosting":
            model.fit(X_train, y_train, sample_weight=sample_weights)
        else:
            model.fit(X_train, y_train)
        
        y_pred = model.predict(X_test)
        f1 = f1_score(y_test, y_pred, average='macro')  
        f1_scores[name] = f1
        print(f"{name} Macro F1 Score: {f1:.4f}")
    
    # Convert to DataFrame and sort by F1 score
    df_f1 = pd.DataFrame(list(f1_scores.items()), columns=['Model', 'F1 Score'])
    df_f1 = df_f1.sort_values(by='F1 Score', ascending=False)
    
    # Plot the F1 scores
    plt.figure(figsize=(12, 6))
    ax = sns.barplot(x='F1 Score', y='Model', data=df_f1, palette='viridis')
    
    for index, value in enumerate(df_f1["F1 Score"]):
        ax.text(value + 0.01, index, f"{value:.2f}", va='center', fontsize=12, fontweight='bold', color='black')
    
    plt.xlabel("F1 Score")
    plt.ylabel("Model")
    plt.title("Comparison of Model Macro F1 Scores for " + y_col)
    plt.xlim(0, 1)
    plt.grid(axis='x', linestyle='--', alpha=0.7)
    plt.show()
    
    return df_f1
```

```python
def class_models_macro_f1_params_balanced(data_inp, X_cols, y_col, test_size, model_params):
    data = data_inp.copy()
    
    # Encode sector
    le = preprocessing.LabelEncoder()
    le.fit(data.Sector)
    data.Sector = le.transform(data.Sector)  
    
    # Split dataset into training and testing while maintaining category proportions
    X = data[X_cols]  # Features
    y = data[y_col]  # Target
    
    X_train, X_test, y_train, y_test = train_test_split(X, y, 
                    test_size=test_size, random_state=123)
    
    
    f1_scores = {}
    
    # Train and evaluate each model using provided parameters
    for name, params in model_params.items():
        model = params["model"](**params["params"])
        
        model.fit(X_train, y_train)
        
        y_pred = model.predict(X_test)
        f1 = f1_score(y_test, y_pred, average='macro')  
        f1_scores[name] = f1
        print(f"{name} Macro F1 Score: {f1:.4f}")
    
    # Convert to DataFrame and sort by F1 score
    df_f1 = pd.DataFrame(list(f1_scores.items()), columns=['Model', 'F1 Score'])
    df_f1 = df_f1.sort_values(by='F1 Score', ascending=False)
    
    # Plot the F1 scores
    plt.figure(figsize=(12, 6))
    ax = sns.barplot(x='F1 Score', y='Model', data=df_f1, palette='viridis')
    
    for index, value in enumerate(df_f1["F1 Score"]):
        ax.text(value + 0.01, index, f"{value:.2f}", va='center', fontsize=12, fontweight='bold', color='black')
    
    plt.xlabel("F1 Score")
    plt.ylabel("Model")
    plt.title("Comparison of Model Macro F1 Scores for " + y_col)
    plt.xlim(0, 1)
    plt.grid(axis='x', linestyle='--', alpha=0.7)
    plt.show()
    
    return df_f1




def class_models_macro_f1_balanced(data_inp, X_cols, y_col, test_size):
    
    data = data_inp.copy()
    
    le = preprocessing.LabelEncoder()
    le.fit(data.Sector)
    data.Sector = le.transform(data.Sector)  # Encode sector

    
    # Split dataset into training and testing while maintaining category proportions
    X = data[X_cols]  # Features
    y = data[y_col]  # Target
    
    X_train, X_test, y_train, y_test = train_test_split(X, y, 
                    test_size=test_size, random_state=123)
    

    
    # Initialize models
    models = {
        "Gradient Boosting": GradientBoostingClassifier(random_state=123),
        "Random Forest": RandomForestClassifier(n_estimators=200, random_state=123),
        "Linear Discriminant Analysis": LinearDiscriminantAnalysis(),
        "Logistic Regression": LogisticRegression(random_state=123, multi_class='multinomial', solver='newton-cg')
    }
    
   
    f1_scores = {}

    # Train and evaluate each model using MACRO F1
    for name, model in models.items():
        

        model.fit(X_train, y_train)
            

        y_pred = model.predict(X_test)
        
        #Compute macro avg F1
        f1 = f1_score(y_test, y_pred, average='macro')  
        f1_scores[name] = f1
        print(f"{name} Macro F1 Score: {f1:.4f}")

    # Train KNN models with k from 1 to 10 and find the best k
    best_knn_k = 1
    best_knn_f1 = 0

    for k in range(1, 11):
        knn_model = KNeighborsClassifier(n_neighbors=k)
        knn_model.fit(X_train, y_train)
        y_pred_knn = knn_model.predict(X_test)
        
        f1_knn = f1_score(y_test, y_pred_knn, average='macro')  
        
        if f1_knn > best_knn_f1:
            best_knn_f1 = f1_knn
            best_knn_k = k
    
    # Store the best KNN Macro F1 score
    f1_scores[f"KNN (k={best_knn_k})"] = best_knn_f1
    print(f"Best KNN Macro F1 Score ({best_knn_k} neighbors): {best_knn_f1:.4f}")

    # Convert to DataFrame and sort by F1 score
    df_f1 = pd.DataFrame(list(f1_scores.items()), columns=['Model', 'F1 Score'])
    df_f1 = df_f1.sort_values(by='F1 Score', ascending=False)

    # Plot the F1 scores
    plt.figure(figsize=(12, 6))
    ax = sns.barplot(x='F1 Score', y='Model', data=df_f1, palette='viridis')
    
    # Add percentage labels on bars
    for index, value in enumerate(df_f1["F1 Score"]):
        ax.text(value + 0.01, index, f"{value:.2f}", va='center', fontsize=12, fontweight='bold', color='black')

    plt.xlabel("F1 Score")
    plt.ylabel("Model")
    plt.title("Comparison of Model Macro F1 Scores for " + y_col)
    plt.xlim(0, 1)  # F1 score ranges from 0 to 1
    plt.grid(axis='x', linestyle='--', alpha=0.7)
    plt.show()

    return df_f1
```
